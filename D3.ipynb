```python
df_O3 = df[(df['Parameter Name']=='Ozone') & (df['Duration Description']=='1 HOUR')]
df_O3.head()
```

```python
df_O3["Date (Local)"] = pd.to_datetime(df_O3["Date (Local)"])
df_O3["Date (Local)"]
```

```python
plt.figure(figsize=(12, 6))

sns.lineplot(data = df_O3, x="Date (Local)", y="Arithmetic Mean")

plt.xlabel("Date")
plt.ylabel("Daily Average O3(ppm)")
plt.title("Daily Average Hourly O3 in Los Angeles, 2023")
plt.show()
```

```python
ppm = 0.035

highest = df_O3.loc[df_O3["Arithmetic Mean"].idxmax()]

days = (df_O3["Arithmetic Mean"] > ppm).sum()

print(highest["Date (Local)"])
print(days)
```
